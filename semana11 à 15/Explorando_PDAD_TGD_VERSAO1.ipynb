```python
domicilios = pd.read_excel("domicilios.xlsx")
col_maior_num_moradores = domicilios['A01npessoas'].max()
id_maior_num_moradores = domicilios[domicilios['A01npessoas'] == col_maior_num_moradores].index[1]
print(f"O domicílio {id_maior_num_moradores} possui a maior quantidade de moradores identificada, que foi {col_maior_num_moradores}.")
```

```python
col_maior_num_criancas = domicilios['A01ncriancas'].max()
id_maior_num_criancas = domicilios[domicilios['A01ncriancas'] == col_maior_num_criancas].index[0]
print(f"O domicílio {id_maior_num_criancas} possui a maior quantidade de crianças identificada, que foi {col_maior_num_criancas}.")
```
